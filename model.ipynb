```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error as mse, r2_score, mean_absolute_error as mae, root_mean_squared_error as rmse
```

Load the Data Set

```python
df= pd.read_csv("https://raw.githubusercontent.com/SagarChhabriya/data-science/refs/heads/main/datasets/TBD/salary_dataset.csv")
df
```

Processing and Splitting

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
X = df[['Experience Years']]
Y = df['Salary']
```

```python
X_tain, X_test, Y_train, Y_test= train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
X_tain.shape
```

```python
Model Training 
```

```python
Lr= LinearRegression()
Lr.fit(X_tain, Y_train)
Y_pred= Lr.predict(X_test)
```

```python
Model Evaluation 
```

```python
r2_score(Y_test, Y_pred)
```

```python
mse(Y_test, Y_pred)
```

```python
rmse(Y_test, Y_pred)
```

```python
mae(Y_test, Y_pred)
```

Model Serialization

```python
import pickle
with open('model.pkl', 'wb') as file:
    pickle.dump(Lr, file)
```

```python
with open('model.pkl', 'rb') as file:
    model = pickle.load(file)
```

```python
model.predict([[0.6]])
```
